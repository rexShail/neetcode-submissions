```
class Dog:
  def __init__(self):
    self.is_animal = True
  def bark(self):
    print("woof")
  def run(self):
    print("run")
dog = Dog()
```

```
class Animal:
  def __init__(self, name, legs=4, barks=True):
    self.name = name
    self.legs = legs
    self.barks = barks

  def info(self):
    print(f"This is an animal named {self.name}, has {self.legs} legs")
    if self.barks:
      print("and this one barks")
    else:
      print("and it does not bark")
```

```
animal = Animal('Cat', 2, False)
animal.info()
```

```
class Budget:
  def __init__(self, budget):
    self.budget = budget
  def expense(self, amount):
    self.budget = self.budget + amount
    print(f"Budget: {self.budget}")
budget = Budget(100)
budget.expense(500)
```
